### Imports

```python
import time
from pathlib import Path
from typing import List

file_path = Path("E:/Learning/RAG Applications/rag_application/tests/fixtures/cv.pdf")
```

#### Method 01: Current Method

```python
from pypdf import PdfReader

def load(file_path: str | Path) -> List[str]:
        print(f"Loading document from: {file_path}")
        reader = PdfReader(file_path)
        output_pages = []

        for page in reader.pages:
            text = page.extract_text()

            if text:
                output_pages.append(text)
        print("Output pages are ready to be returned.")
        return output_pages

# Load the document
start_time = time.time()
pages_of_content = load(file_path)
end_time = time.time()
print(f"Time taken to load the document: {end_time - start_time:.2f} seconds")

for i, page in enumerate(pages_of_content):
    print(f"Page {i + 1}:\n{page}\n\n")
```
